**mode - append , overwrite , errorifexists, ignore

default mode -> errorifexists

overwrite - it will remove the exisitng data overwrite with new data

append - adding new data , adding a new file

ignore - do nothing**

```python
ddl_str="custid int, name string,  age int"
cust_df=spark.read.schema(ddl_str).csv("/Volumes/izwd37dev/wd37db/rawdatta/csvout")
cust_df.write.mode("ignore").csv("/Volumes/izwd37dev/wd37db/rawdatta/csvout",sep="|",header=True)
```

**write dataframe into serilized file format (parquet ) -- format to byte format

parquet
columnar serilized , intelligent , self describing file format**

```python
ddl_str="custid int, fname string, lname string, age int, profession string"
cust_df=spark.read.schema(ddl_str).csv("/Volumes/izwd37dev/wd37db/rawdatta/custs")
#display(cust_df.take(5))
display(cust_df.take(5))

# write to parquet 

cust_df.write.mode("overwrite").parquet("/Volumes/izwd37dev/wd37db/rawdatta/cust_parquet_out")
# part-00000-tid-7560529016580187698-63729798-a7a0-4f3b-a945-a05958b3b737-144-1.c000.snappy.parquet
```

```python

# read parquet data and create dataframe 


cust_df_par=spark.read.parquet("/Volumes/izwd37dev/wd37db/rawdatta/cust_parquet_out/")
cust_df_par.show(5)
cust_df_par.printSchema()
display(cust_df_par)
```
